# Data Cleaning: Company Employee Dataset


```python
import pandas as pd

# Load the dataset
file_path = '/content/Day11_Messy_Company_Employee_Dataset.csv'
df = pd.read_csv(file_path)

# Initial Inspection
print("Dataset Shape:", df.shape)
print("\nColumn Info:")
display(df.info())
print("\nFirst 5 rows:")
display(df.head())
```

```python
# 1. Identify and Quantify Missing Values
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values[missing_values > 0])

# 2. Check for Duplicate Records
duplicates_count = df.duplicated().sum()
print(f"\nTotal duplicate records: {duplicates_count}")
```

```python
# 3. Data Cleaning

# Remove duplicates
df_cleaned = df.drop_duplicates().copy()

# Example Cleaning Strategy (Adjust based on specific column needs observed in info):
# - For numeric columns with few missing values, use median imputation
# - For categorical columns, use mode or 'Unknown'
# - For unique identifiers, drop rows with missing values

# Filling missing numeric values (if any) with median
numeric_cols = df_cleaned.select_dtypes(include=['number']).columns
df_cleaned[numeric_cols] = df_cleaned[numeric_cols].fillna(df_cleaned[numeric_cols].median())

# Filling categorical values with mode or placeholder
object_cols = df_cleaned.select_dtypes(include=['object']).columns
for col in object_cols:
    df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].mode()[0] if not df_cleaned[col].mode().empty else 'Unknown')

print("Cleaning complete. Current missing values count:", df_cleaned.isnull().sum().sum())
```

```python
# 4. Verification and Export
print("Final Dataset Shape:", df_cleaned.shape)

# Export to CSV
output_path = '/content/Cleaned_Company_Employee_Dataset.csv'
df_cleaned.to_csv(output_path, index=False)
print(f"Cleaned dataset saved to: {output_path}")

display(df_cleaned.head())
```
